```python
#Find Body Mass Index
BMI=float(input("Enter the BMI Index Value:"))
if(BMI<=18.5):
    print("Under Weight")
elif(BMI<=24.9):
    print("Normal Weight")
elif(BMI<=29.9):
    print("Over Weight")
elif(BMI<=34.9):
    print("Obese - Class1")
elif(BMI<=39.9):
    print("Obese - Class2")
else:
    print("Obese - Class3")



```

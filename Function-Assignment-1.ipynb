```python
# Create a function that checks whether the given number is Odd or Even

num=int(input("Enter the number:"))

def OddEven():
    if(num % 2 == 0):
        print(num,"is Even Number")
    else:
        print(num,"is Odd Number")

OddEven()
```

```python
"""Create a function that tells eligibility 
of marriage for male and female 
according to their age limit like 21 
for male and 18 for female using only function()"""

def EligiblityForMarriage():
    gender=input("Enter your Gender:")
    age=int(input("Enter your age:"))
    if(gender=="male"):
        if(age>=21):
           print("He is eligible for Marriage")
        else:
            print("He is not eligible for Marriage")
    elif(gender=="female"):
        if(age>=18):
            print("she is eligible for Marriage")
        else:
            print("she is Not eligible for Marriage")
    else:
        print("Please...Enter the gender correctly")
        

EligiblityForMarriage()



```

```python
#print area and perimeter of triangle using Funtion():

height=float(input("Height: "))
weight=float(input("Weight: "))
             

def AreaofTriangle():
    area=(height*weight/2)
    return area

print("Area of Triangle=",AreaofTriangle())

height1=float(input("Height1: "))
height2=float(input("Weight2: "))
breadth=float(input("Breadth: "))

def PerimeterofTriangle():
    area=(height1+height2+breadth)
    return area

print("Perimeter of Triangle=", PerimeterofTriangle())
```
